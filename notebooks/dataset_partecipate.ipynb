# Dataset partecipate

Explore the original company-level source workbook used to derive the hyperedge propagation parameters.

```python
from pathlib import Path

import pandas as pd
from IPython.display import display

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "data").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_PATH = PROJECT_ROOT / "data" / "raw" / "dataset_partecipate.xlsx"
DATA_PATH
```

```python
dataset_partecipate = pd.read_excel(DATA_PATH, sheet_name="Sheet1")

print(f"Rows: {len(dataset_partecipate):,}")
print(f"Columns: {dataset_partecipate.shape[1]:,}")
display(dataset_partecipate.head(20))
```

## Column names

```python
display(pd.DataFrame({"column": dataset_partecipate.columns}))
```

## Columns used for the company propagation parameter

```python
parameter_source_columns = [
    "anno",
    "codfisc",
    "denominazione_imp",
    "fatturato",
    "fatturato_mef",
    "attivo",
    "va",
    "dipendenti",
    "occ_tot_inps",
    "ndip_bil",
    "addetti_mef",
]

missing_columns = [
    column for column in parameter_source_columns
    if column not in dataset_partecipate.columns
]
if missing_columns:
    raise KeyError(f"Missing expected columns: {missing_columns}")

display(dataset_partecipate[parameter_source_columns].head(20))
```
